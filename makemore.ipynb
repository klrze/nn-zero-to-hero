## **Bigrams in Dataset**
Sequence of two adjacent elements

```python
b = {} # initializes empty dictionary to store every unique bigram tuple (as key) and its frequency (as value)

for w in words: # iterates up until indicated # word in list
    # adds start and end tokens to each word in list
    # without this, the model would not know when to start or stop generating a word
    chs = ['<S>'] + list(w) + ['<E>'] 

    # zip(): pairs elements with matching indexes from OG and modified lift
    # assigns ch1 and ch2 to corresponding word (index) in each list
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2) # creates tuple of each bigram (pair of characters)
        b[bigram] = b.get(bigram, 0) + 1 # adds bigram to dictionary and increments its frequency by 1
        print(ch1, ch2) 
```

```python
# takes sequence and rearranges it in ascending order based on the frequency of each bigram
# kv[1] is the metric used by sorted() to determine the order of the bigrams in the dictionary
sorted(b.items(), key=lambda kv: kv[1])
```

## **Counting Bigrams in 2D Torch Tensor**

```python
import torch
```

```python
N = torch.zeros((28, 28), dtype=torch.int32) # initializes a 28x28 tensor of zeros to store the frequency of each bigram
```

```python
# combines names into one string, extracts unique characters, and sorts them in alphabetical order
chars = sorted(list(set(''.join(words)))) 
# maps each char to a unique integer index
stoi = {s:i for i,s in enumerate(chars)} # enumerate: allows loop while keeping track of postion index 
# adds start and end tokens to the dictionary of unique characters
stoi['<S>'] = 26
stoi['<E>'] = 27
```

```python

for w in words:
    chs = ['<S>'] + list(w) + ['<E>'] 
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2] 
        N[ix1, ix2] += 1
```
